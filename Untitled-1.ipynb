```python
"""
Complete pipeline: Add cuisine/meal_type tags → Generate embeddings → Build FAISS
Integrates comprehensive Indian-aware classification
"""

import json
import numpy as np
from tqdm import tqdm
import re
from sentence_transformers import SentenceTransformer
import torch
import faiss
from datetime import datetime
import time

# ============================================================
# COMPREHENSIVE MEAL TYPE KEYWORDS (Indian-aware)
# ============================================================

MEAL_TYPE_KEYWORDS = {
    'breakfast': {
        'strong': [  # High confidence breakfast
            'breakfast', 'brunch', 'morning',
            # Western breakfast
            'pancake', 'waffle', 'cereal', 'oatmeal', 'bagel', 'toast', 
            'eggs', 'omelet', 'omelette', 'frittata', 'quiche', 'scrambled',
            'poached eggs', 'fried eggs', 'bacon', 'sausage', 'hash brown',
            'french toast', 'crepe', 'muffin', 'scone', 'croissant',
            # Indian breakfast
            'dosa', 'idli', 'vada', 'upma', 'poha', 'paratha', 'puri',
            'uttapam', 'appam', 'pesarattu', 'medu vada', 'masala dosa',
            'rava dosa', 'set dosa', 'aloo paratha', 'methi paratha',
            # Other breakfast
            'congee', 'porridge', 'granola', 'smoothie bowl'
        ],
        'ingredients': ['egg', 'eggs', 'bacon', 'sausage', 'oatmeal', 'cereal']
    },
    
    'lunch': {
        'strong': [
            'lunch', 'luncheon',
            # Sandwiches & wraps
            'sandwich', 'wrap', 'panini', 'sub', 'hoagie', 'burger',
            'quesadilla', 'taco', 'burrito', 'pita',
            # Salads & soups
            'salad', 'soup', 'bowl', 'buddha bowl',
            # Indian lunch
            'thali', 'dal rice', 'curd rice', 'rasam rice',
            'chapati', 'roti', 'phulka'
        ],
        'ingredients': ['bread', 'lettuce', 'tomato']
    },
    
    'dinner': {
        'strong': [
            'dinner', 'supper', 'evening',
            # Main courses
            'casserole', 'roast', 'steak', 'chops', 'ribs', 'pot roast',
            'lasagna', 'pasta bake', 'braised', 'stew', 'grill',
            # Indian dinner
            'biryani', 'pulao', 'curry', 'korma', 'vindaloo', 'tikka masala',
            'butter chicken', 'dal makhani', 'paneer', 'tandoori',
            'rogan josh', 'nihari', 'haleem', 'kebab',
            # Other cuisines
            'risotto', 'paella', 'tagine'
        ],
        'ingredients': ['beef', 'pork', 'lamb', 'steak', 'chicken breast']
    },
    
    'snack': {
        'strong': [
            'snack', 'appetizer', 'starter', 'finger food',
            'dip', 'chips', 'bites', 'poppers', 'nachos', 'wings',
            'fingers', 'balls', 'tots', 'fries',
            # Indian snacks
            'samosa', 'pakora', 'bhaji', 'vada pav', 'pani puri',
            'bhel puri', 'sev puri', 'dahi puri', 'chaat', 'kachori',
            'murukku', 'chakli', 'namkeen', 'mixture', 'bonda',
            'cutlet', 'kebab', 'tikka'
        ],
        'ingredients': ['chips', 'crackers']
    },
    
    'dessert': {
        'strong': [
            'dessert', 'sweet', 'pudding',
            # Western desserts
            'cake', 'cookie', 'pie', 'brownie', 'tart', 'cupcake',
            'ice cream', 'sorbet', 'gelato', 'mousse', 'cheesecake',
            'truffle', 'fudge', 'candy', 'chocolate', 'caramel',
            'tiramisu', 'panna cotta',
            # Indian desserts
            'gulab jamun', 'jalebi', 'ladoo', 'barfi', 'halwa',
            'kheer', 'payasam', 'rasgulla', 'ras malai', 'kulfi',
            'shrikhand', 'basundi', 'peda', 'sandesh', 'mysore pak'
        ],
        'ingredients': ['chocolate', 'sugar', 'vanilla extract', 'frosting', 'condensed milk']
    }
}

# ============================================================
# CUISINE RECOGNITION (Comprehensive)
# ============================================================

CUISINE_KEYWORDS = {
    'indian': {
        'dishes': [
            # Breakfast
            'dosa', 'idli', 'vada', 'upma', 'poha', 'paratha', 'puri', 'uttapam',
            # Main
            'biryani', 'pulao', 'curry', 'korma', 'vindaloo', 'tikka', 'tandoori',
            'dal', 'sambar', 'rasam', 'paneer', 'butter chicken', 'rogan josh',
            'chole', 'rajma', 'palak paneer', 'kadai', 'masala',
            # Snacks
            'samosa', 'pakora', 'bhaji', 'chaat', 'pani puri', 'vada pav',
            # Desserts
            'gulab jamun', 'jalebi', 'ladoo', 'barfi', 'halwa', 'kheer', 'kulfi',
            # Breads
            'naan', 'roti', 'chapati', 'kulcha', 'bhatura'
        ],
        'ingredients': [
            'garam masala', 'turmeric', 'cumin', 'coriander', 'cardamom',
            'curry leaves', 'mustard seeds', 'fenugreek', 'asafoetida', 'hing',
            'ghee', 'paneer', 'basmati', 'dal', 'chana', 'besan'
        ],
        'tags': ['indian', 'south indian', 'north indian', 'punjabi', 'mughlai']
    },
    
    'chinese': {
        'dishes': ['fried rice', 'chow mein', 'kung pao', 'sweet and sour', 
                   'stir fry', 'lo mein', 'wonton', 'dumpling', 'dim sum'],
        'ingredients': ['soy sauce', 'oyster sauce', 'sesame oil', 'rice vinegar', 'hoisin'],
        'tags': ['chinese', 'asian']
    },
    
    'italian': {
        'dishes': ['pasta', 'pizza', 'risotto', 'lasagna', 'spaghetti', 'ravioli',
                   'carbonara', 'alfredo', 'marinara', 'pesto', 'gnocchi'],
        'ingredients': ['parmesan', 'mozzarella', 'basil', 'oregano', 'marinara'],
        'tags': ['italian']
    },
    
    'mexican': {
        'dishes': ['taco', 'burrito', 'enchilada', 'quesadilla', 'fajita', 
                   'nachos', 'guacamole', 'salsa', 'tamale'],
        'ingredients': ['cilantro', 'jalapeño', 'cumin', 'chili powder', 'lime'],
        'tags': ['mexican', 'tex-mex']
    },
    
    'thai': {
        'dishes': ['pad thai', 'tom yum', 'green curry', 'red curry', 'thai basil'],
        'ingredients': ['fish sauce', 'lemongrass', 'coconut milk', 'thai basil', 'galangal'],
        'tags': ['thai', 'asian']
    },
    
    'japanese': {
        'dishes': ['sushi', 'sashimi', 'ramen', 'teriyaki', 'tempura', 'miso'],
        'ingredients': ['miso', 'dashi', 'mirin', 'nori', 'wasabi'],
        'tags': ['japanese', 'asian']
    },
    
    'mediterranean': {
        'dishes': ['hummus', 'falafel', 'shawarma', 'kebab', 'gyro', 'tzatziki', 'tabbouleh'],
        'ingredients': ['tahini', 'chickpeas', 'olive oil', 'lemon', 'garlic', 'feta'],
        'tags': ['mediterranean', 'middle eastern', 'greek']
    },
    
    'american': {
        'dishes': ['burger', 'hot dog', 'bbq', 'mac and cheese', 'fried chicken',
                   'meatloaf', 'pot roast', 'clam chowder'],
        'ingredients': ['american cheese', 'barbecue sauce', 'ranch'],
        'tags': ['american', 'southern', 'bbq']
    }
}

# ============================================================
# CLASSIFICATION FUNCTIONS
# ============================================================

def classify_meal_types(recipe: dict) -> list:
    """
    Classify recipe into multiple possible meal types.
    Returns list of applicable meal types.
    """
    title = (recipe.get('title') or '').lower()
    ingredients = [ing.lower() for ing in recipe.get('cleaned_ingredients', []) if ing]
    ingredients_text = ' '.join(ingredients)
    
    scores = {
        'breakfast': 0,
        'lunch': 0,
        'dinner': 0,
        'snack': 0,
        'dessert': 0
    }
    
    # Score each meal type
    for meal_type, keywords in MEAL_TYPE_KEYWORDS.items():
        # Strong keywords in title (weight: 5)
        for keyword in keywords['strong']:
            if keyword in title:
                scores[meal_type] += 5
                break  # Only count once per meal type
        
        # Ingredient keywords (weight: 1)
        for keyword in keywords.get('ingredients', []):
            if keyword in ingredients_text:
                scores[meal_type] += 1
    
    # Special rules
    # Desserts are obvious
    if scores['dessert'] >= 5:
        return ['dessert']
    
    # Determine applicable meal types (threshold-based)
    applicable_meals = []
    
    # If high score, definitely that meal
    for meal_type, score in scores.items():
        if score >= 5:
            applicable_meals.append(meal_type)
    
    # If no strong matches, use softer thresholds
    if not applicable_meals:
        for meal_type, score in scores.items():
            if score >= 2:
                applicable_meals.append(meal_type)
    
    # Special cases for flexible dishes
    # Eggs can be breakfast OR dinner (unless specifically stated)
    if any(ing in ['egg', 'eggs'] for ing in ingredients):
        if 'breakfast' not in applicable_meals and 'dinner' not in title:
            if 'breakfast' not in applicable_meals:
                applicable_meals.append('breakfast')
    
    # Indian dishes can be breakfast, lunch, OR dinner
    indian_breakfast = ['dosa', 'idli', 'vada', 'upma', 'poha', 'paratha', 'puri']
    if any(dish in title for dish in indian_breakfast):
        if 'breakfast' not in applicable_meals:
            applicable_meals.append('breakfast')
        if 'lunch' not in applicable_meals:
            applicable_meals.append('lunch')
        if 'dinner' not in applicable_meals and any(dish in ['dosa', 'idli'] for dish in title.split()):
            applicable_meals.append('dinner')
    
    # Sandwiches/wraps can be lunch OR dinner
    if any(word in title for word in ['sandwich', 'wrap', 'burger']):
        if 'lunch' not in applicable_meals:
            applicable_meals.append('lunch')
        if 'dinner' not in applicable_meals:
            applicable_meals.append('dinner')
    
    # Salads can be lunch OR dinner OR snack
    if 'salad' in title:
        for meal in ['lunch', 'dinner']:
            if meal not in applicable_meals:
                applicable_meals.append(meal)
    
    # Fallback
    if not applicable_meals:
        # Default based on ingredients
        if any(ing in ['beef', 'pork', 'lamb', 'steak'] for ing in ingredients):
            return ['dinner']
        return ['dinner']  # Default fallback
    
    return applicable_meals

def detect_cuisines(recipe: dict) -> list:
    """
    Detect all applicable cuisines for a recipe.
    Returns list of cuisine tags.
    """
    title = (recipe.get('title') or '').lower()
    ingredients = [ing.lower() for ing in recipe.get('cleaned_ingredients', []) if ing]
    ingredients_text = ' '.join(ingredients)
    
    detected_cuisines = set()
    
    for cuisine, keywords in CUISINE_KEYWORDS.items():
        score = 0
        
        # Check dishes (strong signal)
        for dish in keywords['dishes']:
            if dish in title:
                score += 5
                break
        
        # Check ingredients
        for ingredient in keywords['ingredients']:
            if ingredient in ingredients_text:
                score += 1
        
        # If score high enough, add cuisine
        if score >= 3:
            detected_cuisines.update(keywords['tags'])
    
    return sorted(list(detected_cuisines))

# ============================================================
# STEP 1: Add Comprehensive Tags
# ============================================================

def add_comprehensive_tags(metadata_path):
    """Add meal_type (multiple), cuisine, and enhanced search keywords"""
    print("="*70)
    print("STEP 1: Adding Comprehensive Tags (Meal Type + Cuisine)")
    print("="*70)
    
    print(f"\nLoading metadata from {metadata_path}...")
    with open(metadata_path, 'r', encoding='utf-8') as f:
        recipes = json.load(f)
    
    print(f"Total recipes: {len(recipes):,}")
    
    # Create backup
    backup_path = metadata_path.replace('.json', '_BEFORE_TAGS.json')
    print(f"Creating backup at {backup_path}...")
    with open(backup_path, 'w', encoding='utf-8') as f:
        json.dump(recipes, f, indent=2, ensure_ascii=False)
    
    # Statistics
    meal_type_counts = {
        'breakfast': 0, 'lunch': 0, 'dinner': 0, 'snack': 0, 'dessert': 0
    }
    multi_meal_count = 0
    cuisine_counts = {}
    indian_count = 0
    
    print("\nClassifying meal types and cuisines...")
    skipped_count = 0
    
    for recipe in tqdm(recipes, desc="Processing"):
        # Skip recipes with missing critical data
        if not recipe.get('title') or not recipe.get('cleaned_ingredients'):
            skipped_count += 1
            # Set defaults for skipped recipes
            recipe['meal_type'] = ['dinner']
            recipe['cuisine'] = []
            continue
        
        # Classify meal types (multiple)
        meal_types = classify_meal_types(recipe)
        recipe['meal_type'] = meal_types  # Store as list
        
        # Count
        for meal_type in meal_types:
            meal_type_counts[meal_type] += 1
        
        if len(meal_types) > 1:
            multi_meal_count += 1
        
        # Detect cuisines
        cuisines = detect_cuisines(recipe)
        if cuisines:
            recipe['cuisine'] = cuisines
            for cuisine in cuisines:
                cuisine_counts[cuisine] = cuisine_counts.get(cuisine, 0) + 1
            
            if 'indian' in cuisines:
                indian_count += 1
        
        # Update search keywords
        keywords = set(recipe.get('search_keywords', []))
        keywords.update(meal_types)  # Add all meal types
        if cuisines:
            keywords.update(cuisines)  # Add all cuisines
        recipe['search_keywords'] = sorted(list(keywords))
    
    # Save
    print(f"\nSaving updated metadata to {metadata_path}...")
    with open(metadata_path, 'w', encoding='utf-8') as f:
        json.dump(recipes, f, indent=2, ensure_ascii=False)
    
    # Print statistics
    print("\n✓ Comprehensive tagging completed!")
    
    if skipped_count > 0:
        print(f"\n⚠ Skipped {skipped_count:,} recipes with missing data ({skipped_count/len(recipes)*100:.1f}%)")
    
    print(f"\nMeal type distribution:")
    for meal_type, count in sorted(meal_type_counts.items()):
        percentage = (count / len(recipes)) * 100
        print(f"  • {meal_type.capitalize()}: {count:,} ({percentage:.1f}%)")
    
    print(f"\nMulti-meal recipes: {multi_meal_count:,} ({multi_meal_count/len(recipes)*100:.1f}%)")
    
    if cuisine_counts:
        print(f"\nCuisine distribution:")
        for cuisine, count in sorted(cuisine_counts.items(), key=lambda x: x[1], reverse=True):
            percentage = (count / len(recipes)) * 100
            print(f"  • {cuisine.capitalize()}: {count:,} ({percentage:.1f}%)")
        
        print(f"\n🇮🇳 Indian recipes: {indian_count:,} ({indian_count/len(recipes)*100:.1f}%)")
    
    # Show samples
    print("\n" + "="*70)
    print("SAMPLE CLASSIFICATIONS:")
    print("="*70)
    
    # Multi-meal example
    multi_meal = next((r for r in recipes if len(r.get('meal_type', [])) > 1), None)
    if multi_meal:
        print(f"\nMULTI-MEAL RECIPE:")
        print(f"  Title: {multi_meal['title']}")
        print(f"  Meal types: {', '.join(multi_meal['meal_type'])}")
        print(f"  Cuisine: {', '.join(multi_meal.get('cuisine', ['N/A']))}")
    
    # Indian recipe example
    indian_recipe = next((r for r in recipes if 'indian' in r.get('cuisine', [])), None)
    if indian_recipe:
        print(f"\nINDIAN RECIPE:")
        print(f"  Title: {indian_recipe['title']}")
        print(f"  Meal types: {', '.join(indian_recipe['meal_type'])}")
        print(f"  Cuisine: {', '.join(indian_recipe.get('cuisine', []))}")
        print(f"  Ingredients: {', '.join(indian_recipe['cleaned_ingredients'][:5])}")
    
    return recipes

# ============================================================
# STEP 2: Generate Embeddings
# ============================================================

def generate_embeddings(metadata_path, output_path, batch_size=64):
    """Generate embeddings from metadata"""
    print("\n" + "="*70)
    print("STEP 2: Generating Recipe Embeddings")
    print("="*70)
    
    print(f"\nLoading metadata from {metadata_path}...")
    with open(metadata_path, 'r', encoding='utf-8') as f:
        recipes = json.load(f)
    
    print(f"Total recipes: {len(recipes):,}")
    
    # Load model
    print(f"\nLoading embedding model: sentence-transformers/all-MiniLM-L6-v2")
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Using device: {device}")
    
    model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
    model.to(device)
    
    # Create text representations
    print("\nCreating text representations...")
    texts = []
    for recipe in tqdm(recipes, desc="Processing recipes"):
        title = recipe.get('title') or 'Untitled Recipe'
        ingredients = recipe.get('cleaned_ingredients', [])
        # Combine title and ingredients for rich representation
        text = f"{title}. " + " ".join([ing for ing in ingredients if ing])
        texts.append(text)
    
    # Generate embeddings in batches
    print(f"\nGenerating embeddings (batch_size={batch_size})...")
    all_embeddings = []
    
    for i in tqdm(range(0, len(texts), batch_size), desc="Embedding batches"):
        batch = texts[i:i+batch_size]
        embeddings = model.encode(
            batch,
            convert_to_numpy=True,
            show_progress_bar=False,
            device=device
        )
        all_embeddings.append(embeddings)
    
    # Combine all embeddings
    embeddings_array = np.vstack(all_embeddings)
    
    print(f"\nEmbeddings shape: {embeddings_array.shape}")
    print(f"Embedding dimension: {embeddings_array.shape[1]}")
    
    # Save embeddings
    print(f"Saving embeddings to {output_path}...")
    np.save(output_path, embeddings_array)
    
    print(f"\n✓ Embeddings generated and saved!")
    print(f"  • Total vectors: {len(embeddings_array):,}")
    print(f"  • Vector dimension: {embeddings_array.shape[1]}")
    
    return embeddings_array

# ============================================================
# STEP 3: Build FAISS Index
# ============================================================

def build_faiss_index(vectors_path, metadata_path, index_path, ids_path):
    """Build FAISS index from vectors"""
    print("\n" + "="*70)
    print("STEP 3: Building FAISS Index")
    print("="*70)
    
    print(f"\nLoading vectors from {vectors_path}...")
    vectors = np.load(vectors_path, allow_pickle=True)
    
    print(f"Loading metadata from {metadata_path}...")
    with open(metadata_path, 'r', encoding='utf-8') as f:
        recipes = json.load(f)
    
    print(f"\nVectors shape: {vectors.shape}")
    print(f"Total recipes: {len(recipes):,}")
    
    # Verify alignment
    if len(vectors) != len(recipes):
        raise ValueError(f"Mismatch: {len(vectors)} vectors but {len(recipes)} recipes!")
    
    # Extract recipe IDs
    recipe_ids = np.array([recipe['recipe_id'] for recipe in recipes])
    
    # Normalize vectors for cosine similarity
    print("\nNormalizing vectors for cosine similarity...")
    faiss.normalize_L2(vectors)
    
    # Get vector dimension
    dimension = vectors.shape[1]
    
    # Build FAISS index
    print(f"\nBuilding FAISS index (dimension={dimension})...")
    
    if len(vectors) > 100000:
        print("Large dataset detected. Using IVF index for faster search...")
        nlist = min(int(np.sqrt(len(vectors))), 4096)
        quantizer = faiss.IndexFlatIP(dimension)
        index = faiss.IndexIVFFlat(quantizer, dimension, nlist, faiss.METRIC_INNER_PRODUCT)
        
        print(f"Training index with {nlist} clusters...")
        index.train(vectors)
        print("Adding vectors to index...")
        index.add(vectors)
        
        # Set search parameters
        index.nprobe = 10
    else:
        print("Using flat index for exact search...")
        index = faiss.IndexFlatIP(dimension)
        print("Adding vectors to index...")
        index.add(vectors)
    
    print(f"\nIndex contains {index.ntotal:,} vectors")
    
    # Save index
    print(f"Saving FAISS index to {index_path}...")
    faiss.write_index(index, index_path)
    
    # Save recipe IDs
    print(f"Saving recipe IDs to {ids_path}...")
    np.save(ids_path, recipe_ids)
    
    print(f"\n✓ FAISS index built successfully!")
    print(f"  • Index file: {index_path}")
    print(f"  • IDs file: {ids_path}")
    print(f"  • Total indexed recipes: {index.ntotal:,}")
    
    # Test search
    print("\n--- Testing Search ---")
    test_vector = vectors[0:1]
    k = 5
    distances, indices = index.search(test_vector, k)
    
    print(f"Test query: Recipe '{recipes[0]['title']}'")
    print(f"Top {k} similar recipes:")
    for i, (dist, idx) in enumerate(zip(distances[0], indices[0]), 1):
        recipe = recipes[idx]
        cuisine_info = f" [{', '.join(recipe.get('cuisine', ['N/A']))}]"
        meal_info = f" ({', '.join(recipe.get('meal_type', ['N/A']))})"
        print(f"  {i}. {recipe['title']}{cuisine_info}{meal_info} (similarity: {dist:.3f})")
    
    return index, recipe_ids

# ============================================================
# MAIN PIPELINE
# ============================================================

def run_pipeline():
    """Execute complete pipeline"""
    start_time = time.time()
    
    print("\n" + "╔" + "="*68 + "╗")
    print("║" + " "*6 + "COMPREHENSIVE TAGGING + EMBEDDINGS + FAISS PIPELINE" + " "*7 + "║")
    print("╚" + "="*68 + "╝")
    print(f"\nStarted at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
    
    # File paths
    metadata_path = 'claude_recipes_metadata_FINAL.json'
    vectors_path = 'claude_recipe_vectors_FINAL.npy'
    index_path = 'recipe_index.faiss'
    ids_path = 'recipe_ids_FINAL.npy'
    
    try:
        # Step 1: Add comprehensive tags
        recipes = add_comprehensive_tags(metadata_path)
        
        # Step 2: Generate embeddings
        embeddings = generate_embeddings(metadata_path, vectors_path, batch_size=64)
        
        # Step 3: Build FAISS index
        index, recipe_ids = build_faiss_index(vectors_path, metadata_path, index_path, ids_path)
        
        # Success summary
        elapsed = time.time() - start_time
        hours = int(elapsed // 3600)
        minutes = int((elapsed % 3600) // 60)
        seconds = int(elapsed % 60)
        
        print("\n" + "="*70)
        print("🎉 PIPELINE COMPLETED SUCCESSFULLY!")
        print("="*70)
        print(f"\nGenerated/Updated files:")
        print(f"  ✓ {metadata_path} (updated with comprehensive tags)")
        print(f"  ✓ {metadata_path.replace('.json', '_BEFORE_TAGS.json')} (backup)")
        print(f"  ✓ {vectors_path} - {len(embeddings):,} vectors")
        print(f"  ✓ {index_path} - {index.ntotal:,} indexed")
        print(f"  ✓ {ids_path} - {len(recipe_ids):,} IDs")
        
        print(f"\nTotal time: {hours}h {minutes}m {seconds}s")
        print(f"Finished at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print("\n✨ Your recipe recommendation system is ready!")
        print("   Now supports: Multi-meal types, Indian cuisine, and comprehensive tags!")
        
        return True
        
    except Exception as e:
        print(f"\n❌ PIPELINE FAILED!")
        print(f"Error: {str(e)}")
        import traceback
        traceback.print_exc()
        return False

if __name__ == "__main__":
    import sys
    success = run_pipeline()
    sys.exit(0 if success else 1)
```

```python
import torch
import numpy as np
import json
import os

# --- CONFIG ---
NPY_PATH = "/data1/home/sathvik/Documents/FRESH/claude_recipe_vectors_FINAL.npy"
META_PATH = "/data1/home/sathvik/Documents/FRESH/claude_recipes_metadata_FINAL.json"
OUTPUT_DIR = "/data1/home/sathvik/Documents/FRESH"

print(f"📂 Loading data from {NPY_PATH}...")

# 1. LOAD & CONVERT VECTORS
# Load NumPy file
numpy_vectors = np.load(NPY_PATH)

# Convert to PyTorch Tensor
recipe_vectors = torch.from_numpy(numpy_vectors).float()
print(f"✅ Converted {len(recipe_vectors)} recipes to PyTorch Tensors.")

# Save as 'recipe_vectors.pt' so the main script can find it
torch.save(recipe_vectors, os.path.join(OUTPUT_DIR, "recipe_vectors.pt"))
print(f"💾 Saved 'recipe_vectors.pt' (Size: {recipe_vectors.shape})")

# 2. GENERATE MACRO ANCHORS (The Missing Piece)
print("🧪 Generating Macro Anchors from Metadata...")

with open(META_PATH, "r") as f:
    meta_raw = json.load(f)
    
# Handle List vs Dict format
if isinstance(meta_raw, list):
    metadata = {str(r.get("recipe_id") or r.get("id")): r for r in meta_raw}
    recipe_ids = [str(r.get("recipe_id") or r.get("id")) for r in meta_raw]
else:
    metadata = meta_raw
    recipe_ids = list(metadata.keys())

# We need to find 3 real recipes to act as "North Stars" for nutrition
anchors = {
    "Protein": None, # Will find a Chicken/Egg recipe
    "Carbs": None,   # Will find a Rice/Pasta recipe
    "Fats": None     # Will find a Oil/Cream recipe
}

# Simple Keyword Search to find anchors
for i, rid in enumerate(recipe_ids):
    if i >= len(recipe_vectors): break
    
    ingredients = str(metadata[rid].get("cleaned_ingredients", "")).lower()
    
    # Find a strong Protein candidate
    if anchors["Protein"] is None and ("chicken breast" in ingredients or "egg white" in ingredients):
        anchors["Protein"] = recipe_vectors[i]
        print(f"   Found Protein Anchor: {metadata[rid].get('title')}")
        
    # Find a strong Carb candidate
    if anchors["Carbs"] is None and ("rice" in ingredients or "pasta" in ingredients or "potato" in ingredients):
        anchors["Carbs"] = recipe_vectors[i]
        print(f"   Found Carb Anchor: {metadata[rid].get('title')}")

    # Find a strong Fat candidate
    if anchors["Fats"] is None and ("butter" in ingredients or "cream" in ingredients or "oil" in ingredients):
        anchors["Fats"] = recipe_vectors[i]
        print(f"   Found Fat Anchor: {metadata[rid].get('title')}")

# Fallback: If we didn't find specific ones, just use random vectors (rare case)
if anchors["Protein"] is None: anchors["Protein"] = recipe_vectors[0]
if anchors["Carbs"] is None: anchors["Carbs"] = recipe_vectors[1]
if anchors["Fats"] is None: anchors["Fats"] = recipe_vectors[2]

# Save the Macros file
torch.save(anchors, os.path.join(OUTPUT_DIR, "macro_vectors.pt"))
print("💾 Saved 'macro_vectors.pt'")
print("\n✅ SETUP COMPLETE. You can now run 'python train_distilled.py'")
```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer
import torch
import numpy as np

def preprocess_2m_dataset():
    # 1. Load JSON Data
    print("Loading 2.2M JSON Dataset...")
    # Update this path to your actual file location
    json_path = "/data1/home/sathvik/Documents/FRESH/claude_recipes_metadata_FINAL.json"
    
    try:
        df = pd.read_json(json_path)
    except ValueError:
        print("Standard load failed, trying lines=True...")
        df = pd.read_json(json_path, lines=True)

    print(f"Loaded {len(df)} recipes.")
    print(f"Columns found: {list(df.columns)}")

    # 2. Data Cleaning & Formatting
    # We define which columns contain lists that need string conversion
    # Your dataset uses 'search_keywords', 'diet', 'meal_type', etc.
    list_cols = ['ingredients', 'search_keywords', 'diet', 'meal_type', 'cuisine']
    
    def clean_field(val):
        if isinstance(val, list):
            # Join lists with comma
            return ", ".join([str(v) for v in val])
        if pd.isna(val):
            return ""
        return str(val)

    print("Cleaning list columns...")
    for col in list_cols:
        if col in df.columns:
            df[col] = df[col].apply(clean_field)
        else:
            print(f"⚠️ Warning: '{col}' missing. Creating empty column.")
            df[col] = ""

    # 3. Create Rich Semantic Text for SBERT
    # We use 'search_keywords' + 'cuisine' + 'diet' as the "Tags" context
    print("Creating combined text...")
    df['combined_text'] = (
        "Title: " + df['title'].astype(str) + ". " + 
        "Cuisine: " + df['cuisine'] + ". " +
        "Diet: " + df['diet'] + ". " +
        "Ingredients: " + df['ingredients'] + ". " + 
        "Keywords: " + df['search_keywords']
    )
    
    # 4. Encode Semantic Vectors
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Encoding Recipes on {device}...")
    
    model = SentenceTransformer('all-MiniLM-L6-v2', device=device)
    
    # Batch size 512 is usually safe for 384-dim models on standard GPUs (16GB+ VRAM)
    # If OOM error, reduce to 256
    embeddings = model.encode(
        df['combined_text'].tolist(), 
        show_progress_bar=True, 
        batch_size=512, 
        convert_to_numpy=True
    )
    
    # 5. Save Embeddings
    print("Saving Embeddings to .pt file...")
    torch.save(torch.tensor(embeddings), "recipe_embeddings_2m.pt")
    
    # 6. Save Metadata for Training
    # IMPORTANT: We save the columns that exist in YOUR dataset
    # Renaming 'diet' -> 'diet_labels' logic handled here if your training script expects 'diet_labels'
    print("Saving Metadata to .pkl file...")
    
    # We explicitly select useful columns for the 'Fresh_Real_Dataset' class
    meta_df = df[['title', 'ingredients', 'diet', 'cuisine', 'search_keywords']].copy()
    
    # If your training script strictly asks for 'tags' and 'diet_labels', we rename them here:
    meta_df.rename(columns={
        'diet': 'diet_labels', 
        'search_keywords': 'tags'
    }, inplace=True)
    
    meta_df.to_pickle("recipe_metadata_2m.pkl")
    
    print("✅ Data Processed! Files created:")
    print("   - recipe_embeddings_2m.pt")
    print("   - recipe_metadata_2m.pkl")

if __name__ == "__main__":
    preprocess_2m_dataset()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import numpy as np
from tqdm import tqdm
import time
import os

# ==============================================================================
# 1. THE REAL DATASET LOADER (Uses your 2.2M processed files)
# ==============================================================================
class FRESH_Real_Dataset(Dataset):
    def __init__(self, embeddings_path, metadata_path):
        print(f"Loading Embeddings from {embeddings_path}...")
        # Load tensors to CPU RAM (approx 3.5GB for 2.2M float32 vectors)
        self.recipe_vecs = torch.load(embeddings_path, map_location='cpu') 
        
        print(f"Loading Metadata from {metadata_path}...")
        self.metadata = pd.read_pickle(metadata_path)
        
        self.num_samples = len(self.recipe_vecs)
        print(f"✅ Loaded {self.num_samples} items for training.")

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        # --- A. POSITIVE SAMPLE (The Anchor) ---
        pos_recipe_vec = self.recipe_vecs[idx]
        
        # Simulate a Perfect User Match (User Vector ≈ Recipe Vector)
        # We add slight noise so the model learns robustness, not just identity mapping
        user_vec = pos_recipe_vec + torch.randn(384) * 0.1
        
        # Simulate a Relevant Health Constraint
        # (e.g., if Recipe is "Salad", Constraint is "Low Calorie")
        constraint_vec = pos_recipe_vec + torch.randn(384) * 0.1

        # Positive Labels: [Pantry=1.0, Safety=1.0]
        pos_explicit = torch.tensor([1.0, 1.0]) 
        pos_align = torch.tensor([0.95])  
        pos_label = torch.tensor([1.0])

        # --- B. HARD NEGATIVE MINING (The SOTA Secret) ---
        # We pick a random recipe to serve as the "Bad Option"
        rand_idx = np.random.randint(0, self.num_samples)
        neg_recipe_vec = self.recipe_vecs[rand_idx]
        
        # Decide HOW it fails (Pantry vs. Health)
        fail_type = np.random.choice(['pantry', 'health'])
        
        if fail_type == 'pantry':
            # FAIL: Perfect semantics/health, but MISSING INGREDIENTS
            # We use the POSITIVE recipe vector but flag it as "Pantry Missing"
            # This teaches the model: "Even if it tastes great, if I don't have ingredients, Score = 0"
            neg_train_vec = pos_recipe_vec.clone() 
            neg_explicit = torch.tensor([0.1, 1.0]) # [Low Pantry, Safe]
            neg_align = torch.tensor([0.95])        # High Align
            
        elif fail_type == 'health':
            # FAIL: Perfect taste/pantry, but UNSAFE DIET
            # We use the POSITIVE recipe vector but flag it as "Unsafe"
            neg_train_vec = pos_recipe_vec.clone()
            neg_explicit = torch.tensor([1.0, 0.0]) # [High Pantry, Unsafe]
            neg_align = torch.tensor([0.1])         # Force Low Align score constraint
            
        neg_label = torch.tensor([0.0])

        return {
            'user_vec': user_vec,
            'constraint_vec': constraint_vec,
            'pos_recipe': pos_recipe_vec,
            'pos_explicit': pos_explicit,
            'pos_align': pos_align,
            'pos_label': pos_label,
            'neg_recipe': neg_train_vec,
            'neg_explicit': neg_explicit,
            'neg_align': neg_align,
            'neg_label': neg_label
        }

# ==============================================================================
# 2. MODEL ARCHITECTURE (FRESH Network)
# ==============================================================================
class FeatureInteractionLayer(nn.Module):
    def __init__(self, dim, num_heads=4, dropout=0.1):
        super().__init__()
        self.multihead_attn = nn.MultiheadAttention(embed_dim=dim, num_heads=num_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(dim)

    def forward(self, x):
        attn_out, _ = self.multihead_attn(x, x, x)
        return self.norm(x + attn_out)

class FRESH_Network(nn.Module):
    def __init__(self, embedding_dim=384):
        super(FRESH_Network, self).__init__()
        # TOWER 1: Taste
        self.taste_encoder = nn.Sequential(
            nn.Linear(embedding_dim * 2, 256),
            nn.LayerNorm(256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, 128)
        )
        # TOWER 2: Constraints
        self.constraint_encoder = nn.Sequential(
            nn.Linear(embedding_dim, 256),
            nn.LayerNorm(256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, 128)
        )
        # TOWER 3: Wide Features
        self.wide_encoder = nn.Sequential(nn.Linear(2, 16), nn.ReLU())
        
        # Interaction
        self.interaction = FeatureInteractionLayer(dim=128, num_heads=4)
        
        # Head
        self.head = nn.Sequential(
            nn.Linear(128 * 2 + 16 + 1, 64), 
            nn.LayerNorm(64), nn.GELU(),
            nn.Linear(64, 1), nn.Sigmoid() 
        )

    def forward(self, recipe_vecs, user_vecs, constraint_vecs, explicit_feats, align_score):
        taste_input = torch.cat([recipe_vecs, user_vecs], dim=1) 
        taste_emb = self.taste_encoder(taste_input)
        constraint_emb = self.constraint_encoder(constraint_vecs)
        
        seq_input = torch.stack([taste_emb, constraint_emb], dim=1)
        interacted = self.interaction(seq_input)
        
        flat_features = interacted.view(interacted.size(0), -1)
        wide_out = self.wide_encoder(explicit_feats)
        
        final_input = torch.cat([flat_features, wide_out, align_score], dim=1)
        return self.head(final_input)

# ==============================================================================
# 3. TRAINING ENGINE
# ==============================================================================
def train():
    # SETTINGS
    EPOCHS = 5
    BATCH_SIZE = 64
    LR = 1e-4
    EMBED_PATH = "recipe_embeddings_2m.pt"  # Verify this filename
    META_PATH = "recipe_metadata_2m.pkl"    # Verify this filename

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"🚀 Starting Training on {device}")

    # Load Data
    dataset = FRESH_Real_Dataset(EMBED_PATH, META_PATH)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=4)

    # Init Model
    model = FRESH_Network().to(device)
    optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
    criterion = nn.BCELoss()

    # Training Loop
    for epoch in range(EPOCHS):
        model.train()
        total_loss = 0
        start_time = time.time()
        
        progress = tqdm(dataloader, desc=f"Epoch {epoch+1}/{EPOCHS}")
        
        for batch in progress:
            # Move to GPU
            user_vecs = batch['user_vec'].to(device)
            constraint_vecs = batch['constraint_vec'].to(device)
            
            # 1. Forward Pass (Positive)
            pos_preds = model(
                recipe_vecs=batch['pos_recipe'].to(device),
                user_vecs=user_vecs,
                constraint_vecs=constraint_vecs,
                explicit_feats=batch['pos_explicit'].to(device),
                align_score=batch['pos_align'].to(device)
            )
            pos_loss = criterion(pos_preds, batch['pos_label'].to(device))

            # 2. Forward Pass (Negative)
            neg_preds = model(
                recipe_vecs=batch['neg_recipe'].to(device),
                user_vecs=user_vecs,
                constraint_vecs=constraint_vecs,
                explicit_feats=batch['neg_explicit'].to(device),
                align_score=batch['neg_align'].to(device)
            )
            neg_loss = criterion(neg_preds, batch['neg_label'].to(device))

            # 3. Update
            loss = pos_loss + neg_loss
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            progress.set_postfix(loss=loss.item())

        print(f"✅ Epoch {epoch+1} Done. Loss: {total_loss/len(dataloader):.4f}. Time: {time.time()-start_time:.0f}s")
        
        # Save Checkpoint
        torch.save(model.state_dict(), f"fresh_model_epoch_{epoch+1}.pth")

    print("🎉 Training Finished!")

if __name__ == "__main__":
    train()
```

```python
import torch
import pandas as pd
import numpy as np
from model import FRESH_Network   # Ensure this imports your model class

class FRESH_Inference_Engine:
    def __init__(self, model_path="fresh_model_epoch_5.pth", metadata_path="recipe_metadata_2m.pkl", embed_path="recipe_embeddings_2m.pt"):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"🚀 Initializing FRESH Engine on {self.device}...")
        
        # 1. Load the Model Skeleton
        self.model = FRESH_Network().to(self.device)
        
        # 2. Load the Trained Weights (The "Brain")
        print(f"Loading weights from {model_path}...")
        state_dict = torch.load(model_path, map_location=self.device)
        self.model.load_state_dict(state_dict)
        self.model.eval() # Set to Evaluation Mode (Freezes Dropout, etc.)
        
        # 3. Load Data for Retrieval
        print("Loading Recipe Database...")
        self.metadata = pd.read_pickle(metadata_path)
        self.recipe_vecs = torch.load(embed_path, map_location=self.device)
        print("✅ FRESH Engine Ready!")

    def predict(self, user_vec, constraint_vec, candidate_indices, pantry_match_scores, diet_safe_flags):
        """
        Scores a list of candidate recipes against the User and Constraint.
        """
        with torch.no_grad():
            # Prepare Inputs
            # 1. Get Recipe Vectors for candidates
            batch_recipe_vecs = self.recipe_vecs[candidate_indices]
            
            # 2. Expand User/Constraint vectors to match batch size
            batch_size = len(candidate_indices)
            batch_user_vecs = user_vec.expand(batch_size, -1)
            batch_constraint_vecs = constraint_vec.expand(batch_size, -1)
            
            # 3. Prepare Explicit Features [Pantry%, Safety]
            # Shape: [Batch, 2]
            explicit_feats = torch.stack([
                torch.tensor(pantry_match_scores, device=self.device),
                torch.tensor(diet_safe_flags, device=self.device)
            ], dim=1)
            
            # 4. Calculate Alignment Score (Cosine Sim)
            # Dot product between Recipe and Constraint
            align_scores = torch.nn.functional.cosine_similarity(
                batch_recipe_vecs, batch_constraint_vecs
            ).unsqueeze(1) # Shape: [Batch, 1]

            # 5. RUN THE MODEL
            scores = self.model(
                recipe_vecs=batch_recipe_vecs,
                user_vecs=batch_user_vecs,
                constraint_vecs=batch_constraint_vecs,
                explicit_feats=explicit_feats,
                align_score=align_scores
            )
            
            return scores.squeeze().cpu().numpy()

# --- USAGE EXAMPLE ---
if __name__ == "__main__":
    # Test loading
    engine = FRESH_Inference_Engine()
    print("Test Load Successful.")
```
